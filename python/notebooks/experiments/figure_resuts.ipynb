```python
import numpy as np

def pad_cm_to_3x3(cm_2x2):
    cm3 = np.zeros((3,3), dtype=int)
    cm3[:2, :2] = cm_2x2
    return cm3
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

def plot_cm_triplet(
    cms, titles, tick_labels,
    legend_texts, legend_ncol,
    outpath=None,
    figsize=(9, 4),
    margins=dict(left=0.06, right=0.96, top=0.86, bottom=0.22, wspace=0.28),
    cmap="Blues"
):
    """
    Fixed-geometry 1x3 confusion-matrix figure (consistent across different class counts).
    """
    fig, axes = plt.subplots(1, 3, figsize=figsize)

    # FIXED margins -> same final footprint for all figures
    fig.subplots_adjust(**margins)

    vmin = 0
    vmax = max(cm.max() for cm in cms)

    im = None
    for ax, cm, title in zip(axes, cms, titles):
        im = ax.imshow(cm, cmap=cmap, vmin=vmin, vmax=vmax)
        ax.grid(False)
        ax.set_title(title, fontsize=18, pad=14)

        n = cm.shape[0]
        ax.set_xticks(range(n))
        ax.set_yticks(range(n))
        ax.set_xticklabels(tick_labels, fontsize=16, fontweight="bold")
        ax.set_yticklabels(tick_labels, fontsize=16, fontweight="bold")
        ax.tick_params(length=0, pad=7)

        # Force square axes boxes (consistent look)
        try:
            ax.set_box_aspect(1)  # Matplotlib >= 3.3
        except Exception:
            ax.set_aspect("equal", adjustable="box")

        # Annotate counts
        for i in range(n):
            for j in range(n):
                ax.text(j, i, f"{cm[i, j]}",
                        ha="center", va="center",
                        fontsize=18,
                        color="white" if cm[i, j] > vmax * 0.5 else "black")

    axes[0].set_ylabel("True", fontsize=20, labelpad=4)
    axes[1].set_xlabel("Predicted", fontsize=20, labelpad=4)
    # Reserve enough space for legend + xlabel
    fig.subplots_adjust(left=0.10,
                        right=0.90,
                        top=0.86,
                        bottom=0.34,
                        wspace=0.10)
    legend_handles = [Patch(facecolor="none", edgecolor="none", label=t) for t in legend_texts]

    fig.legend(handles=legend_handles,
            loc="lower center",
            bbox_to_anchor=(0.5, 0.01),   # push legend down inside the reserved bottom margin
            ncol=legend_ncol,
            frameon=False,
            fontsize=18,
            columnspacing=0.01,
            handletextpad=0.01)
    
    if outpath is not None:
        fig.savefig(outpath, dpi=300, bbox_inches="tight")
    return fig, axes
```

```python
CM_RF  = np.array([[65, 3, 0],
                   [ 0, 2, 0],
                   [ 0, 0, 5]])

CM_SVM = np.array([[65, 1, 2],
                   [ 0, 2, 0],
                   [ 0, 0, 5]])

CM_KNN = np.array([[65, 2, 1],
                   [ 0, 2, 0],
                   [ 0, 0, 5]])

plot_cm_triplet(
    cms=[CM_RF, CM_SVM, CM_KNN],
    titles=["RF", "SVM", "KNN"],
    tick_labels=["0", "A", "C"],
    legend_texts=["0: Healthy", "A: Auxiliary Leakage", "C: Combined Leakage"],
    legend_ncol=3,
    outpath="cm_multiclass.png"
)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Patch

# --- Confusion matrices ---
CM_RF  = np.array([[65, 3, 0],
                   [ 0, 2, 0],
                   [ 0, 0, 5]])

CM_SVM = np.array([[65, 1, 2],
                   [ 0, 2, 0],
                   [ 0, 0, 5]])

CM_KNN = np.array([[65, 2, 1],
                   [ 0, 2, 0],
                   [ 0, 0, 5]])

cms = [CM_RF, CM_SVM, CM_KNN]
titles = ["RF", "SVM", "KNN"]
tick_labels = ["0", "A", "C"]

fig, axes = plt.subplots(1, 3, figsize=(12, 4))
plt.subplots_adjust(wspace=0.28)

vmin = 0
vmax = max(cm.max() for cm in cms)

for ax, cm, title in zip(axes, cms, titles):

    im = ax.imshow(cm, cmap="Blues", vmin=vmin, vmax=vmax)
    ax.grid(False)

    ax.set_title(title, fontsize=18, pad=12)

    ax.set_xticks(range(3))
    ax.set_yticks(range(3))
    ax.set_xticklabels(tick_labels, fontsize=16, fontweight="bold")
    ax.set_yticklabels(tick_labels, fontsize=16, fontweight="bold")

    ax.tick_params(length=0, pad=7)

    # Annotate counts
    for i in range(3):
        for j in range(3):
            ax.text(j, i, f"{cm[i, j]}",
                    ha="center", va="center",
                    fontsize=18,
                    color="white" if cm[i, j] > vmax*0.5 else "black")

axes[0].set_ylabel("True", fontsize=16)
axes[1].set_xlabel("Predicted", fontsize=16)

# Shared colorbar
# cbar = fig.colorbar(im, ax=axes, fraction=0.03, pad=0.02)
# cbar.set_label("Count", fontsize=15)
# cbar.ax.tick_params(labelsize=13)

# --- Legend mapping ---
legend_elements = [
    Patch(facecolor="none", edgecolor="none", label="0: Healthy"),
    Patch(facecolor="none", edgecolor="none", label="A: Auxiliary Leakage"),
    Patch(facecolor="none", edgecolor="none", label="C: Combined Leakage"),
]

fig.legend(handles=legend_elements,
           loc="lower center",
           ncol=3,
           frameon=False,
           fontsize=20,
           columnspacing=0.3,
           handletextpad=0.1,
           borderaxespad=0.1)

# Reserve space at bottom for legend
plt.tight_layout(rect=[0.1, 0.1, 0.9, 1])
```
